```python
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
import pickle
import os
```

```python
# 1. Kyle primitives
# ============================================================
def kyle_step(beta, lam, v, p_prev, u):
    x = beta * (v - p_prev)
    y = x + u
    p_next = p_prev + lam * y
    u_ins = x * (v - p_next)
    u_mm  = -(p_next - v) ** 2
    return x, y, p_next, u_ins, u_mm

def bayes_update_sigma(Sigma_prev, beta, sigma_u):
    denom = beta**2 * Sigma_prev + sigma_u**2
    if denom < 1e-12:
        return Sigma_prev
    return Sigma_prev * sigma_u**2 / denom
```

```python
# 2. Discretization (CRUCIAL)
# ============================================================
def q(x, step):
    return round(x / step) * step

def insider_infoset(v, p_prev, sigma_v, sigma_u, Sigma_prev, t,T):
    return (
        "I",
        q(v, 0.25),
        q(p_prev, 0.05),
        q(sigma_v, 0.5),
        q(sigma_u, 0.5),
        q(Sigma_prev, 0.5),
        round(t / T, 2)
    )

def mm_infoset(y, sigma_v, sigma_u, Sigma_prev, t, T):
    return (
        "M",
        q(y, 0.25),
        q(sigma_v, 0.5),
        q(sigma_u, 0.5),
        q(Sigma_prev, 0.5),
        round(t / T, 2)
    )
```

```python
# 3. Regret matching
# ============================================================
def regret_matching(r):
    r_pos = np.maximum(r, 0.0)
    s = r_pos.sum()
    if s > 1e-12:
        return r_pos / s
    return np.ones_like(r) / len(r)
```

```python
def train_cfr(
    T,
    num_episodes,
    INSIDER_ACTIONS,
    MM_ACTIONS,
    P0,
    sigma_v,
    sigma_u,
    seed=0,
    snapshot_interval=None,
    snapshot_dir='cfr_snapshots'
):
    rng = np.random.default_rng(seed)

    R_I = defaultdict(lambda: np.zeros(len(INSIDER_ACTIONS)))
    R_M = defaultdict(lambda: np.zeros(len(MM_ACTIONS)))
    S_I = defaultdict(lambda: np.zeros(len(INSIDER_ACTIONS)))
    S_M = defaultdict(lambda: np.zeros(len(MM_ACTIONS)))

    realized_xp = [[] for _ in range(T)]

    if snapshot_interval is not None:
        os.makedirs(snapshot_dir, exist_ok=True)

    for ep in range(num_episodes):
        v = rng.normal(P0, sigma_v)
        u_seq = rng.normal(0, sigma_u, T)

        p = P0
        Sigma = sigma_v**2
        traj = []

        # ---------- forward simulation ----------
        for t in range(T):
            sI = insider_infoset(v, p, sigma_v, sigma_u, Sigma, t, T)
            piI = regret_matching(R_I[sI])
            S_I[sI] += piI
            i_idx = rng.choice(len(INSIDER_ACTIONS), p=piI)
            beta = INSIDER_ACTIONS[i_idx]

            u = u_seq[t]
            x = beta * (v - p)
            y = x + u

            sM = mm_infoset(y, sigma_v, sigma_u, Sigma, t, T)
            piM = regret_matching(R_M[sM])
            S_M[sM] += piM
            j_idx = rng.choice(len(MM_ACTIONS), p=piM)
            lam = MM_ACTIONS[j_idx]

            x, y, p_next, uI, uM = kyle_step(beta, lam, v, p, u)
            realized_xp[t].append((x, p_next))

            traj.append((sI, sM, beta, lam, uI, uM, p, u, Sigma))
            Sigma = bayes_update_sigma(Sigma, beta, sigma_u)
            p = p_next

        # ---------- backward regret update ----------
        cont_I = np.zeros(T + 1)
        cont_M = np.zeros(T + 1)
        for t in reversed(range(T)):
            cont_I[t] = traj[t][4] + cont_I[t + 1]
            cont_M[t] = traj[t][5] + cont_M[t + 1]

        for t in range(T):
            sI, sM, beta, lam, _, _, p_prev, u, Sigma_prev = traj[t]
            vI = cont_I[t]
            vM = cont_M[t]

            # Insider deviations
            for i, b_alt in enumerate(INSIDER_ACTIONS):
                _, _, _, uI_alt, _ = kyle_step(b_alt, lam, v, p_prev, u)
                R_I[sI][i] += (uI_alt + cont_I[t + 1] - vI)

            # MM deviations
            for j, l_alt in enumerate(MM_ACTIONS):
                _, _, _, _, uM_alt = kyle_step(beta, l_alt, v, p_prev, u)
                R_M[sM][j] += (uM_alt + cont_M[t + 1] - vM)

        # ---------- Take snapshot ----------
        if snapshot_interval is not None and (ep + 1) % snapshot_interval == 0:
            snapshot_path = os.path.join(snapshot_dir, f'snapshot_ep_{ep+1}.pkl')
            print(f"Saving snapshot at episode {ep+1} to {snapshot_path}")

            current_avg_I = {}
            for s in S_I:
                s_sum = S_I[s].sum()
                if s_sum > 1e-12:
                    current_avg_I[s] = S_I[s] / s_sum
                else:
                    current_avg_I[s] = np.ones_like(S_I[s]) / len(S_I[s]) # Default to uniform if no visits

            current_avg_M = {}
            for s in S_M:
                s_sum = S_M[s].sum()
                if s_sum > 1e-12:
                    current_avg_M[s] = S_M[s] / s_sum
                else:
                    current_avg_M[s] = np.ones_like(S_M[s]) / len(S_M[s]) # Default to uniform if no visits

            snapshot_data = {
                'avg_I': current_avg_I,
                'avg_M': current_avg_M,
                'realized_xp': realized_xp,
                'episode': ep + 1
            }
            with open(snapshot_path, 'wb') as f:
                pickle.dump(snapshot_data, f)

    final_avg_I = {}
    for s in S_I:
        s_sum = S_I[s].sum()
        if s_sum > 1e-12:
            final_avg_I[s] = S_I[s] / s_sum
        else:
            final_avg_I[s] = np.ones_like(S_I[s]) / len(S_I[s])

    final_avg_M = {}
    for s in S_M:
        s_sum = S_M[s].sum()
        if s_sum > 1e-12:
            final_avg_M[s] = S_M[s] / s_sum
        else:
            final_avg_M[s] = np.ones_like(S_M[s]) / len(S_M[s])

    return final_avg_I, final_avg_M, realized_xp
```

```python
# 5. Correct probability heatmaps
# ============================================================
def plot_joint_heatmaps(realized_xp, T):
    periods = [0, T//4, T // 2, T - 1]
    fig, axs = plt.subplots(1, 4, figsize=(20, 6))
    fig.suptitle("Joint Probability of (x, p) under CFR", fontsize=16)

    for ax, t in zip(axs, periods):
        data = realized_xp[t]
        x, p = zip(*data)

        hist, xe, pe = np.histogram2d(x, p, bins=50)
        prob = hist / hist.sum()

        im = ax.imshow(
            prob.T,
            origin="lower",
            aspect="auto",
            extent=[xe[0], xe[-1], pe[0], pe[-1]],
            cmap="viridis"
        )
        fig.colorbar(im, ax=ax, label="Probability")
        ax.set_title(f"Period {t+1}")
        ax.set_xlabel("Insider quantity x")
        ax.set_ylabel("Price p")

    plt.tight_layout()
    plt.show()
```

```python
# 6. Run
# ============================================================
if __name__ == "__main__":
    P0 = 0.5
    SIGMA_V = 2.0
    SIGMA_U = 1.0
    T = 25
    EPISODES = 50_000
    SNAPSHOT_INTERVAL = 10000
    SNAPSHOT_DIR = 'cfr_snapshots'

    INSIDER_ACTIONS = np.linspace(0.0, 2.0, 9)   # beta
    MM_ACTIONS = np.linspace(0.2, 0.6, 9)        # lambda

    print(f"Starting CFR training for {EPISODES} episodes...")
    avg_I, avg_M, final_realized_xp = train_cfr(
        T=T,
        num_episodes=EPISODES,
        INSIDER_ACTIONS=INSIDER_ACTIONS,
        MM_ACTIONS=MM_ACTIONS,
        P0=P0,
        sigma_v=SIGMA_V,
        sigma_u=SIGMA_U,
        seed=0,
        snapshot_interval=SNAPSHOT_INTERVAL,
        snapshot_dir=SNAPSHOT_DIR
    )
    print("CFR training complete. Now plotting snapshots...")

    # Plot heatmaps for each snapshot
    snapshot_files = sorted([f for f in os.listdir(SNAPSHOT_DIR) if f.startswith('snapshot_ep_') and f.endswith('.pkl')])

    for filename in snapshot_files:
        filepath = os.path.join(SNAPSHOT_DIR, filename)
        with open(filepath, 'rb') as f:
            snapshot_data = pickle.load(f)

        print(f"Plotting for episode: {snapshot_data['episode']}")
        plot_joint_heatmaps(snapshot_data['realized_xp'], T)

    # Optionally, plot the final state as well if it wasn't exactly at an interval
    # if EPISODES % SNAPSHOT_INTERVAL != 0:
    #     print(f"Plotting final state after {EPISODES} episodes.")
    #     plot_joint_heatmaps(final_realized_xp, T)
```
