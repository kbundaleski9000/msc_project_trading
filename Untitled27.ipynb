```python
"""
Kyle model CFR / MC-CFR implementation (end-to-end) aligned with professor remarks,
including:

✅ Order & pedagogy
1) Intuition: CFR = DFS recursion to leaves; regrets drive policy; output = average strategy.
2) Vanilla CFR on a tiny discrete Kyle (exact chance enumeration).
3) MC-CFR (External Sampling) for larger Kyle (continuous chance; tabular infosets).
4) Evaluation: treat learned CFR policies like RL policies (stochastic). Sample trajectories and compare
   to multi-period Kyle equilibrium sequences (Beta_n, Lambda_n, Sigma_n).

✅ Key algorithmic corrections
- External Sampling MC-CFR regret updates: DO NOT multiply by opponent reach when sampling opponent from its own policy.
  (So regrets update as: R[I,a] += (u(a) - V(I)) without reach_M/reach_I factors.)
- Average strategy accumulation uses OWN reach.

✅ Tiny Kyle evaluation in SAME STYLE as MC-CFR evaluation
- Plot learned CFR actions vs equilibrium(-proxy) actions against time t.
  For tiny discretized Kyle, equilibrium is taken as analytic continuous Kyle equilibrium projected onto the discrete grids.

✅ Extra: Tiny Kyle convergence diagnostics
- Positive regret mass over iterations (tracked infosets)
- Action probabilities over iterations (tracked infosets)

✅ Proactive experimentation hook (prof's meta-comment)
- Toggle include_p_in_mm_infoset for MC-CFR; run both and record MAE/plots.

Notes:
- Tiny Kyle is a discretized toy (general-sum), so "equilibrium" there is a PROXY:
  analytic continuous Kyle equilibrium projected onto discrete action sets.
"""

import os
import pickle
from collections import defaultdict
from dataclasses import dataclass
from typing import Dict, Tuple, Any, Optional, List

import numpy as np
import matplotlib.pyplot as plt

# Optional: if SciPy is available we use brentq; otherwise we fall back to bisection.
try:
    from scipy.optimize import brentq  # type: ignore
    _HAVE_SCIPY = True
except Exception:
    brentq = None
    _HAVE_SCIPY = False


# ============================================================
# Utilities
# ============================================================

def regret_matching(regrets: np.ndarray) -> np.ndarray:
    r_pos = np.maximum(regrets, 0.0)
    s = float(r_pos.sum())
    if s > 1e-15:
        return r_pos / s
    return np.ones_like(regrets) / len(regrets)


def safe_normalize(x: np.ndarray) -> np.ndarray:
    s = float(x.sum())
    if s > 1e-15:
        return x / s
    return np.ones_like(x) / len(x)


def bisect_root(f, a: float, b: float, tol: float = 1e-10, max_iter: int = 200) -> float:
    """
    Simple bisection root finder on (a,b) where f(a)*f(b) < 0.
    Used if SciPy isn't available.
    """
    fa = f(a)
    fb = f(b)
    if not np.isfinite(fa) or not np.isfinite(fb) or fa * fb > 0:
        raise ValueError("Bisection requires a bracket with opposite signs and finite values.")
    lo, hi = a, b
    flo, fhi = fa, fb
    for _ in range(max_iter):
        mid = 0.5 * (lo + hi)
        fmid = f(mid)
        if not np.isfinite(fmid):
            hi = mid
            fhi = fmid
            continue
        if abs(fmid) < tol or (hi - lo) < tol:
            return mid
        if flo * fmid <= 0:
            hi, fhi = mid, fmid
        else:
            lo, flo = mid, fmid
    return 0.5 * (lo + hi)


def nearest_in_grid(x: float, grid: np.ndarray) -> float:
    idx = int(np.argmin(np.abs(grid - float(x))))
    return float(grid[idx])


def var_of_three_point_symmetric(a: float) -> float:
    # Values are {-a, 0, +a} each with prob 1/3
    # Var = E[X^2] = (a^2 + 0 + a^2)/3 = 2a^2/3
    return (2.0 * a * a) / 3.0


# ============================================================
# Multi-period Kyle equilibrium solver (your algorithm)
# ============================================================

def solve_kyle_model(N: int, sigma2_v: float, sigma2_u: float):
    """
    Two-pass algorithm (backward for h_n, forward for Sigma, Lambda, Beta, Alpha).
    Returns arrays:
      Sigma: length N+1 (Sigma_0..Sigma_N)
      Lambda, Beta, Alpha, h: length N
    """
    sigma_u = float(np.sqrt(sigma2_u))

    # --- PASS 1: BACKWARD iteration for h_n ---
    h = np.zeros(N, dtype=float)
    h[-1] = 0.5  # terminal condition (n = N)

    for n in reversed(range(N - 1)):
        h_next = float(h[n + 1])
        theta_next = float(np.sqrt(h_next / (1.0 - h_next)))

        def root_func(h_n: float) -> float:
            # bracket (0.5, 1.0)
            if (h_n <= 0.5) or (h_n >= 1.0):
                return np.inf
            return theta_next - (2.0 * h_n - 1.0) / (h_n * np.sqrt(1.0 - h_n))

        lo = 0.5 + 1e-9
        hi = 1.0 - 1e-9

        if _HAVE_SCIPY:
            h[n] = float(brentq(root_func, lo, hi))
        else:
            xs = np.linspace(lo, hi, 2000)
            vals = np.array([root_func(x) for x in xs], dtype=float)
            ok = np.isfinite(vals)
            xs, vals = xs[ok], vals[ok]
            idx = None
            for k in range(len(vals) - 1):
                if vals[k] == 0.0:
                    idx = k
                    break
                if vals[k] * vals[k + 1] < 0:
                    idx = k
                    break
            if idx is None:
                raise RuntimeError("Could not bracket root for h_n without SciPy. Install scipy for brentq.")
            a, b = float(xs[idx]), float(xs[idx + 1])
            h[n] = float(bisect_root(root_func, a, b))

    # --- PASS 2: FORWARD iteration for equilibrium sequences ---
    Sigma = np.zeros(N + 1, dtype=float)
    Lambda = np.zeros(N, dtype=float)
    Beta = np.zeros(N, dtype=float)
    Alpha = np.zeros(N, dtype=float)

    Sigma[0] = float(sigma2_v)

    for n in range(N):
        Sigma[n + 1] = float(h[n] * Sigma[n])

        # NOTE: keep consistent with your lecture notes
        Lambda[n] = float(np.sqrt((1.0 - h[n]) * Sigma[n + 1]) / sigma_u)
        Beta[n] = float((np.sqrt(1.0 - h[n]) * sigma_u) / np.sqrt(Sigma[n + 1]))
        Alpha[n] = float((sigma_u / (2.0 * np.sqrt(Sigma[n]))) * np.sqrt(h[n] / (1.0 - h[n])))

    return {"Sigma": Sigma, "Lambda": Lambda, "Beta": Beta, "Alpha": Alpha, "h": h}


# ============================================================
# Tiny discrete Kyle — Vanilla CFR (exact chance enumeration)
# ============================================================

def kyle_step_x_lambda(x: float, lam: float, v: float, p_prev: float, u: float):
    y = x + u
    p_next = p_prev + lam * y
    u_ins = x * (v - p_next)
    u_mm = -(p_next - v) ** 2
    return y, p_next, u_ins, u_mm


@dataclass
class TinyKyleConfig:
    T: int = 2
    P0: float = 0.0
    V: float = 1.0
    N: float = 1.0
    X_ACTIONS: Tuple[float, ...] = (-1.0, 0.0, 1.0)
    LAM_ACTIONS: Tuple[float, ...] = (0.0, 0.5, 1.0)
    seed: int = 0


class TinyKyleVanillaCFR:
    """
    Vanilla CFR on a tiny discrete Kyle game with exact chance enumeration.
    Includes convergence diagnostics and the SAME-style eval plot as MC-CFR:
    learned vs equilibrium-proxy actions over time t.
    """

    def __init__(self, cfg: TinyKyleConfig):
        self.cfg = cfg
        self.X_ACTIONS = np.array(cfg.X_ACTIONS, dtype=float)
        self.LAM_ACTIONS = np.array(cfg.LAM_ACTIONS, dtype=float)

        self.V_vals = np.array([-cfg.V, 0.0, +cfg.V], dtype=float)
        self.U_vals = np.array([-cfg.N, 0.0, +cfg.N], dtype=float)

        self.pV = np.array([1/3, 1/3, 1/3], dtype=float)
        self.pU = np.array([1/3, 1/3, 1/3], dtype=float)

        self.R_I = defaultdict(lambda: np.zeros(len(self.X_ACTIONS)))
        self.R_M = defaultdict(lambda: np.zeros(len(self.LAM_ACTIONS)))
        self.S_I = defaultdict(lambda: np.zeros(len(self.X_ACTIONS)))
        self.S_M = defaultdict(lambda: np.zeros(len(self.LAM_ACTIONS)))

    def infoset_insider(self, t: int, v: float, p_prev: float):
        return ("I", t, float(v), round(float(p_prev), 6))

    def infoset_mm(self, t: int, y: float, p_prev: float):
        return ("M", t, round(float(y), 6), round(float(p_prev), 6))

    def cfr(
        self,
        t: int,
        p_prev: float,
        v: float,
        u_seq: np.ndarray,
        player_to_update: str,
        reach_I: float,
        reach_M: float,
        reach_C: float
    ) -> Tuple[float, float]:
        if t == self.cfg.T:
            return 0.0, 0.0

        u = float(u_seq[t])

        # --- Insider node ---
        I = self.infoset_insider(t, v, p_prev)
        sigma_I = regret_matching(self.R_I[I])

        if player_to_update == "I":
            # avg strategy accumulation: own reach * chance
            self.S_I[I] += (reach_I * reach_C) * sigma_I

        util_ins_if_x = np.zeros(len(self.X_ACTIONS))
        util_mm_if_x = np.zeros(len(self.X_ACTIONS))

        for a_idx, x in enumerate(self.X_ACTIONS):
            y = float(x) + u

            # --- MM node ---
            M = self.infoset_mm(t, y, p_prev)
            sigma_M = regret_matching(self.R_M[M])

            if player_to_update == "M":
                self.S_M[M] += (reach_M * reach_C) * sigma_M

            util_ins_if_lam = np.zeros(len(self.LAM_ACTIONS))
            util_mm_if_lam = np.zeros(len(self.LAM_ACTIONS))

            for b_idx, lam in enumerate(self.LAM_ACTIONS):
                _, p_next, u_ins, u_mm = kyle_step_x_lambda(float(x), float(lam), float(v), float(p_prev), u)
                cont_ins, cont_mm = self.cfr(
                    t + 1, p_next, v, u_seq, player_to_update,
                    reach_I * float(sigma_I[a_idx]),
                    reach_M * float(sigma_M[b_idx]),
                    reach_C
                )
                util_ins_if_lam[b_idx] = u_ins + cont_ins
                util_mm_if_lam[b_idx] = u_mm + cont_mm

            V_ins_at_M = float(sigma_M @ util_ins_if_lam)
            V_mm_at_M = float(sigma_M @ util_mm_if_lam)

            if player_to_update == "M":
                cf_reach_M = reach_I * reach_C
                self.R_M[M] += cf_reach_M * (util_mm_if_lam - V_mm_at_M)

            util_ins_if_x[a_idx] = V_ins_at_M
            util_mm_if_x[a_idx] = V_mm_at_M

        V_ins_at_I = float(sigma_I @ util_ins_if_x)
        V_mm_at_I = float(sigma_I @ util_mm_if_x)

        if player_to_update == "I":
            cf_reach_I = reach_M * reach_C
            self.R_I[I] += cf_reach_I * (util_ins_if_x - V_ins_at_I)

        return V_ins_at_I, V_mm_at_I

    def train(
        self,
        num_passes_over_chance: int = 800,
        track_v: Optional[float] = None,
        track_p0: Optional[float] = None,
    ):
        """
        Exact chance enumeration.
        Returns average strategies and history for convergence diagnostics.
        """
        T = self.cfg.T
        if track_v is None:
            track_v = float(self.cfg.V)
        if track_p0 is None:
            track_p0 = float(self.cfg.P0)

        # Precompute all u-sequences and probs
        u_sequences: List[np.ndarray] = []
        u_probs: List[float] = []

        def rec_build_u(t: int, seq: List[float], prob: float):
            if t == T:
                u_sequences.append(np.array(seq, dtype=float))
                u_probs.append(prob)
                return
            for u_idx, u in enumerate(self.U_vals):
                rec_build_u(t + 1, seq + [float(u)], prob * float(self.pU[u_idx]))

        rec_build_u(0, [], 1.0)

        I_track = self.infoset_insider(0, float(track_v), float(track_p0))
        M_track = self.infoset_mm(0, 0.0, float(track_p0))

        hist = {
            "iter": [],
            "I_pos_regret_sum": [],
            "M_pos_regret_sum": [],
            "I_policy": [],
            "M_policy": [],
        }

        for it in range(1, num_passes_over_chance + 1):
            for v_idx, v in enumerate(self.V_vals):
                p_v = float(self.pV[v_idx])
                for u_seq, p_u in zip(u_sequences, u_probs):
                    reach_C = p_v * float(p_u)
                    self.cfr(0, self.cfg.P0, float(v), u_seq, "I", 1.0, 1.0, reach_C)
                    self.cfr(0, self.cfg.P0, float(v), u_seq, "M", 1.0, 1.0, reach_C)

            # record diagnostics every 10 iterations
            if it % 10 == 0:
                rI = self.R_I[I_track]
                rM = self.R_M[M_track]
                hist["iter"].append(it)
                hist["I_pos_regret_sum"].append(float(np.maximum(rI, 0.0).sum()))
                hist["M_pos_regret_sum"].append(float(np.maximum(rM, 0.0).sum()))
                hist["I_policy"].append(regret_matching(rI).copy())
                hist["M_policy"].append(regret_matching(rM).copy())

        avg_I, avg_M = self.get_avg_strategies()
        return avg_I, avg_M, hist, (I_track, M_track)

    def get_avg_strategies(self):
        avg_I = {I: safe_normalize(s.copy()) for I, s in self.S_I.items()}
        avg_M = {M: safe_normalize(s.copy()) for M, s in self.S_M.items()}
        return avg_I, avg_M





def simulate_tiny_policy_vs_kyle_equilibrium(
    avg_I: Dict[Any, np.ndarray],
    avg_M: Dict[Any, np.ndarray],
    cfg: TinyKyleConfig,
    num_simulations: int = 5000,
    seed: int = 123
):
    """
    Tiny Kyle evaluation in SAME STYLE as MC-CFR evaluation:
    - simulate trajectories under learned avg strategies
    - simulate 'equilibrium' as analytic continuous Kyle equilibrium projected onto discrete grids
    - plot learned avg actions vs equilibrium-proxy actions across time t
    """
    rng = np.random.default_rng(seed)

    X_GRID = np.array(cfg.X_ACTIONS, dtype=float)
    LAM_GRID = np.array(cfg.LAM_ACTIONS, dtype=float)

    # use tiny discrete distribution variances to parameterize the analytic Kyle solver
    sigma2_v = var_of_three_point_symmetric(cfg.V)
    sigma2_u = var_of_three_point_symmetric(cfg.N)

    eq = solve_kyle_model(cfg.T, sigma2_v, sigma2_u)
    beta_eq = eq["Beta"]
    lam_eq = eq["Lambda"]

    avg_x_cfr = np.zeros(cfg.T)
    avg_lam_cfr = np.zeros(cfg.T)
    avg_x_eq = np.zeros(cfg.T)
    avg_lam_eq = np.zeros(cfg.T)

    V_vals = np.array([-cfg.V, 0.0, +cfg.V], dtype=float)
    U_vals = np.array([-cfg.N, 0.0, +cfg.N], dtype=float)

    for _ in range(num_simulations):
        v = float(rng.choice(V_vals))
        u_seq = rng.choice(U_vals, size=cfg.T)

        p = float(cfg.P0)
        p_e = float(cfg.P0)

        for t in range(cfg.T):
            u = float(u_seq[t])

            # --- CFR ---
            I = ("I", t, float(v), round(float(p), 6))
            piI = avg_I.get(I, np.ones(len(X_GRID)) / len(X_GRID))
            x = float(rng.choice(X_GRID, p=piI))

            y = x + u
            M = ("M", t, round(float(y), 6), round(float(p), 6))
            piM = avg_M.get(M, np.ones(len(LAM_GRID)) / len(LAM_GRID))
            lam = float(rng.choice(LAM_GRID, p=piM))

            p_next = p + lam * y

            avg_x_cfr[t] += x
            avg_lam_cfr[t] += lam
            p = p_next

            # --- Equilibrium-proxy (continuous Kyle projected onto discrete action grids) ---
            x_cont = float(beta_eq[t]) * (v - p_e)
            x_eq_t = nearest_in_grid(x_cont, X_GRID)
            lam_eq_t = nearest_in_grid(float(lam_eq[t]), LAM_GRID)

            y_e = x_eq_t + u
            p_next_e = p_e + lam_eq_t * y_e

            avg_x_eq[t] += x_eq_t
            avg_lam_eq[t] += lam_eq_t
            p_e = p_next_e

    avg_x_cfr /= num_simulations
    avg_lam_cfr /= num_simulations
    avg_x_eq /= num_simulations
    avg_lam_eq /= num_simulations

    return {
        "avg_x_cfr": avg_x_cfr,
        "avg_lam_cfr": avg_lam_cfr,
        "avg_x_eq": avg_x_eq,
        "avg_lam_eq": avg_lam_eq,
        "beta_eq_cont": beta_eq,
        "lambda_eq_cont": lam_eq,
    }





# ============================================================
# Large Kyle — MC-CFR (External Sampling)
# ============================================================

def bayes_update_sigma(Sigma_prev: float, beta: float, sigma_u: float) -> float:
    denom = beta**2 * Sigma_prev + sigma_u**2
    if denom < 1e-15:
        return Sigma_prev
    return Sigma_prev * sigma_u**2 / denom


def kyle_step_beta_lambda(beta: float, lam: float, v: float, p_prev: float, u: float):
    x = beta * (v - p_prev)
    y = x + u
    p_next = p_prev + lam * y
    u_ins = x * (v - p_next)
    u_mm = -(p_next - v) ** 2
    return x, y, p_next, u_ins, u_mm


def q(x: float, step: float) -> float:
    return round(float(x) / step) * step


def insider_infoset(v: float, p_prev: float, Sigma_prev: float, t: int, T: int):
    return ("I",
            q(v, 0.25),
            q(p_prev, 0.05),
            q(Sigma_prev, 0.5),
            t)


def mm_infoset(y: float, Sigma_prev: float, t: int, T: int, p_prev: Optional[float] = None):
    if p_prev is None:
        return ("M",
                q(y, 0.25),
                q(Sigma_prev, 0.5),
                t)
    return ("M",
            q(y, 0.25),
            q(p_prev, 0.05),
            q(Sigma_prev, 0.5),
            t)


class KyleExternalSamplingMCCFR:
    """
    MC-CFR (External Sampling) for the continuous-chance Kyle model.

     Key correction:
    Since we sample opponent actions from their current policy, regret updates do NOT multiply by opponent reach.
    Regret update at infoset is:
        R[I, a] += u(a) - V(I)
    """

    def __init__(
        self,
        T: int,
        INSIDER_ACTIONS: np.ndarray,
        MM_ACTIONS: np.ndarray,
        P0: float,
        sigma_v: float,
        sigma_u: float,
        seed: int = 0,
        snapshot_interval: Optional[int] = None,
        snapshot_dir: str = "cfr_snapshots",
        include_p_in_mm_infoset: bool = False,
    ):
        self.T = int(T)
        self.INSIDER_ACTIONS = np.array(INSIDER_ACTIONS, dtype=float)
        self.MM_ACTIONS = np.array(MM_ACTIONS, dtype=float)
        self.P0 = float(P0)
        self.sigma_v = float(sigma_v)
        self.sigma_u = float(sigma_u)
        self.include_p_in_mm_infoset = bool(include_p_in_mm_infoset)

        self.rng = np.random.default_rng(seed)

        self.R_I = defaultdict(lambda: np.zeros(len(self.INSIDER_ACTIONS)))
        self.R_M = defaultdict(lambda: np.zeros(len(self.MM_ACTIONS)))
        self.S_I = defaultdict(lambda: np.zeros(len(self.INSIDER_ACTIONS)))
        self.S_M = defaultdict(lambda: np.zeros(len(self.MM_ACTIONS)))

        self.snapshot_interval = snapshot_interval
        self.snapshot_dir = snapshot_dir
        if self.snapshot_interval is not None:
            os.makedirs(self.snapshot_dir, exist_ok=True)

    def _mm_state(self, y: float, Sigma_prev: float, t: int, p_prev: float):
        if self.include_p_in_mm_infoset:
            return mm_infoset(y, Sigma_prev, t, self.T, p_prev=p_prev)
        return mm_infoset(y, Sigma_prev, t, self.T, p_prev=None)

    def _save_snapshot(self, episode: int, extra: Optional[Dict[str, Any]] = None):
        avg_I = {s: safe_normalize(v.copy()) for s, v in self.S_I.items()}
        avg_M = {s: safe_normalize(v.copy()) for s, v in self.S_M.items()}
        payload = {"avg_I": avg_I, "avg_M": avg_M, "episode": episode}
        if extra:
            payload.update(extra)
        path = os.path.join(self.snapshot_dir, f"snapshot_ep_{episode}.pkl")
        with open(path, "wb") as f:
            pickle.dump(payload, f)
        print(f"[snapshot] saved {path}")

    def _cfr_traverse(
        self,
        t: int,
        p_prev: float,
        Sigma_prev: float,
        v: float,
        u_seq: np.ndarray,
        player_to_update: str,
        reach_I: float,
        reach_M: float,
    ) -> Tuple[float, float]:
        if t == self.T:
            return 0.0, 0.0

        u = float(u_seq[t])

        # ---------- Insider node ----------
        sI = insider_infoset(v, p_prev, Sigma_prev, t, self.T)
        sigma_I = regret_matching(self.R_I[sI])

        # avg strategy accumulation: own reach (chance sampled)
        if player_to_update == "I":
            self.S_I[sI] += reach_I * sigma_I

        if player_to_update == "I":
            util_ins_if_beta = np.zeros(len(self.INSIDER_ACTIONS))
            util_mm_if_beta = np.zeros(len(self.INSIDER_ACTIONS))

            for i, beta in enumerate(self.INSIDER_ACTIONS):
                beta = float(beta)
                x = beta * (v - p_prev)
                y = x + u

                sM = self._mm_state(y, Sigma_prev, t, p_prev)
                sigma_M = regret_matching(self.R_M[sM])

                # sample MM action from its own policy
                j = int(self.rng.choice(len(self.MM_ACTIONS), p=sigma_M))
                lam = float(self.MM_ACTIONS[j])

                _, _, p_next, u_ins, u_mm = kyle_step_beta_lambda(beta, lam, v, p_prev, u)
                Sigma_next = bayes_update_sigma(Sigma_prev, beta, self.sigma_u)

                cont_ins, cont_mm = self._cfr_traverse(
                    t + 1, p_next, Sigma_next, v, u_seq, player_to_update,
                    reach_I * float(sigma_I[i]),
                    reach_M * float(sigma_M[j]),
                )
                util_ins_if_beta[i] = u_ins + cont_ins
                util_mm_if_beta[i] = u_mm + cont_mm

            V_ins = float(sigma_I @ util_ins_if_beta)
            V_mm = float(sigma_I @ util_mm_if_beta)

            # ✅ External sampling regret update (NO opponent reach factor)
            self.R_I[sI] += (util_ins_if_beta - V_ins)

            return V_ins, V_mm

        # sample insider action (opponent node)
        i = int(self.rng.choice(len(self.INSIDER_ACTIONS), p=sigma_I))
        beta = float(self.INSIDER_ACTIONS[i])

        x = beta * (v - p_prev)
        y = x + u

        # ---------- MM node ----------
        sM = self._mm_state(y, Sigma_prev, t, p_prev)
        sigma_M = regret_matching(self.R_M[sM])

        if player_to_update == "M":
            self.S_M[sM] += reach_M * sigma_M

        if player_to_update == "M":
            util_ins_if_lam = np.zeros(len(self.MM_ACTIONS))
            util_mm_if_lam = np.zeros(len(self.MM_ACTIONS))

            for j, lam in enumerate(self.MM_ACTIONS):
                lam = float(lam)

                _, _, p_next, u_ins, u_mm = kyle_step_beta_lambda(beta, lam, v, p_prev, u)
                Sigma_next = bayes_update_sigma(Sigma_prev, beta, self.sigma_u)

                cont_ins, cont_mm = self._cfr_traverse(
                    t + 1, p_next, Sigma_next, v, u_seq, player_to_update,
                    reach_I * float(sigma_I[i]),
                    reach_M * float(sigma_M[j]),
                )
                util_ins_if_lam[j] = u_ins + cont_ins
                util_mm_if_lam[j] = u_mm + cont_mm

            V_ins = float(sigma_M @ util_ins_if_lam)
            V_mm = float(sigma_M @ util_mm_if_lam)

            # ✅ External sampling regret update (NO opponent reach factor)
            self.R_M[sM] += (util_mm_if_lam - V_mm)

            return V_ins, V_mm

        # opponent node: sample MM action
        j = int(self.rng.choice(len(self.MM_ACTIONS), p=sigma_M))
        lam = float(self.MM_ACTIONS[j])

        _, _, p_next, u_ins, u_mm = kyle_step_beta_lambda(beta, lam, v, p_prev, u)
        Sigma_next = bayes_update_sigma(Sigma_prev, beta, self.sigma_u)

        cont_ins, cont_mm = self._cfr_traverse(
            t + 1, p_next, Sigma_next, v, u_seq, player_to_update,
            reach_I * float(sigma_I[i]),
            reach_M * float(sigma_M[j]),
        )
        return u_ins + cont_ins, u_mm + cont_mm

    def train(
        self,
        num_episodes: int = 5_000,
        log_every: int = 1_000,
        eq_for_logging: Optional[Dict[str, np.ndarray]] = None,
        eval_simulations: int = 200,
        eval_seed: int = 123,
    ):
        """
        Train external sampling MC-CFR. Optionally logs MAE vs equilibrium over training.
        """
        mae_hist = {"ep": [], "mae_beta": [], "mae_lambda": []}

        for ep in range(1, num_episodes + 1):
            v = float(self.rng.normal(self.P0, self.sigma_v))
            u_seq = self.rng.normal(0.0, self.sigma_u, self.T)

            self._cfr_traverse(0, self.P0, self.sigma_v**2, v, u_seq,
                               player_to_update="I",
                               reach_I=1.0, reach_M=1.0)

            self._cfr_traverse(0, self.P0, self.sigma_v**2, v, u_seq,
                               player_to_update="M",
                               reach_I=1.0, reach_M=1.0)

            if self.snapshot_interval is not None and (ep % self.snapshot_interval == 0):
                self._save_snapshot(ep, extra={"mae_hist": mae_hist})

            if log_every and (ep % log_every == 0):
                print(f"[train] ep={ep}/{num_episodes}  |I_infosets|={len(self.R_I)}  |M_infosets|={len(self.R_M)}")

                if eq_for_logging is not None:
                    avg_I = {s: safe_normalize(v.copy()) for s, v in self.S_I.items()}
                    avg_M = {s: safe_normalize(v.copy()) for s, v in self.S_M.items()}
                    res = evaluate_policies_vs_multi_period_equilibrium(
                        avg_I=avg_I,
                        avg_M=avg_M,
                        T=self.T,
                        INSIDER_ACTIONS=self.INSIDER_ACTIONS,
                        MM_ACTIONS=self.MM_ACTIONS,
                        P0=self.P0,
                        sigma_v=self.sigma_v,
                        sigma_u=self.sigma_u,
                        eq=eq_for_logging,
                        num_simulations=eval_simulations,
                        seed=eval_seed,
                        include_p_in_mm_infoset=self.include_p_in_mm_infoset
                    )
                    mae_b = float(np.mean(np.abs(res["avg_beta_cfr"] - res["beta_eq"])))
                    mae_l = float(np.mean(np.abs(res["avg_lam_cfr"] - res["lambda_eq"])))
                    mae_hist["ep"].append(ep)
                    mae_hist["mae_beta"].append(mae_b)
                    mae_hist["mae_lambda"].append(mae_l)
                    print(f"        MAE(beta)={mae_b:.4f} | MAE(lambda)={mae_l:.4f}")

        avg_I = {s: safe_normalize(v.copy()) for s, v in self.S_I.items()}
        avg_M = {s: safe_normalize(v.copy()) for s, v in self.S_M.items()}
        return avg_I, avg_M, mae_hist


# ============================================================
# Evaluation / Plotting (MC-CFR vs equilibrium sequences)
# ============================================================

def evaluate_policies_vs_multi_period_equilibrium(
    avg_I: Dict[Any, np.ndarray],
    avg_M: Dict[Any, np.ndarray],
    T: int,
    INSIDER_ACTIONS: np.ndarray,
    MM_ACTIONS: np.ndarray,
    P0: float,
    sigma_v: float,
    sigma_u: float,
    eq: Dict[str, np.ndarray],
    num_simulations: int = 500,
    seed: int = 123,
    include_p_in_mm_infoset: bool = False
):
    """
    Evaluate:
    - average chosen beta_t and lambda_t vs equilibrium Beta[t], Lambda[t]
    - average x_t and p_t trajectories
    """
    rng = np.random.default_rng(seed)

    beta_eq = eq["Beta"]
    lam_eq = eq["Lambda"]
    Sigma_eq_seq = eq["Sigma"]

    avg_beta_cfr = np.zeros(T)
    avg_lam_cfr = np.zeros(T)
    avg_x_cfr = np.zeros(T)
    avg_p_cfr = np.zeros(T)

    avg_x_eq = np.zeros(T)
    avg_p_eq = np.zeros(T)

    for _ in range(num_simulations):
        v = float(rng.normal(P0, sigma_v))
        u_seq = rng.normal(0.0, sigma_u, T)

        p = float(P0)
        Sigma = float(sigma_v**2)

        p_e = float(P0)

        for t in range(T):
            u = float(u_seq[t])

            # --- CFR policy sampling ---
            sI = insider_infoset(v, p, Sigma, t, T)
            piI = avg_I.get(sI, np.ones(len(INSIDER_ACTIONS)) / len(INSIDER_ACTIONS))
            beta = float(rng.choice(INSIDER_ACTIONS, p=piI))

            x = beta * (v - p)
            y = x + u

            if include_p_in_mm_infoset:
                sM = mm_infoset(y, Sigma, t, T, p_prev=p)
            else:
                sM = mm_infoset(y, Sigma, t, T, p_prev=None)

            piM = avg_M.get(sM, np.ones(len(MM_ACTIONS)) / len(MM_ACTIONS))
            lam = float(rng.choice(MM_ACTIONS, p=piM))

            x_val, _, p_next, _, _ = kyle_step_beta_lambda(beta, lam, v, p, u)

            avg_beta_cfr[t] += beta
            avg_lam_cfr[t] += lam
            avg_x_cfr[t] += x_val
            avg_p_cfr[t] += p_next

            Sigma = bayes_update_sigma(Sigma, beta, sigma_u)
            p = p_next

            # --- Equilibrium baseline ---
            beta_t = float(beta_eq[t])
            lam_t = float(lam_eq[t])

            x_e = beta_t * (v - p_e)
            y_e = x_e + u
            p_next_e = p_e + lam_t * y_e

            avg_x_eq[t] += x_e
            avg_p_eq[t] += p_next_e

            # for baseline, keep solver Sigma path (optional)
            _ = float(Sigma_eq_seq[t + 1])
            p_e = p_next_e

    avg_beta_cfr /= num_simulations
    avg_lam_cfr /= num_simulations
    avg_x_cfr /= num_simulations
    avg_p_cfr /= num_simulations
    avg_x_eq /= num_simulations
    avg_p_eq /= num_simulations

    return {
        "avg_beta_cfr": avg_beta_cfr,
        "avg_lam_cfr": avg_lam_cfr,
        "avg_x_cfr": avg_x_cfr,
        "avg_p_cfr": avg_p_cfr,
        "avg_x_eq": avg_x_eq,
        "avg_p_eq": avg_p_eq,
        "beta_eq": beta_eq,
        "lambda_eq": lam_eq,
        "Sigma_eq": Sigma_eq_seq
    }


def plot_eval(results: Dict[str, np.ndarray], title_prefix: str = "MC-CFR vs multi-period Kyle equilibrium"):
    T = len(results["avg_beta_cfr"])
    t = np.arange(1, T + 1)



    plt.figure(figsize=(12, 4))
    plt.title(f"{title_prefix}: avg x_t")
    plt.plot(t, results["avg_x_cfr"], label="CFR avg x_t", linewidth=2)
    plt.plot(t, results["avg_x_eq"], label="Equilibrium avg x_t", linestyle="--", linewidth=2)
    plt.xlabel("t")
    plt.ylabel("x")
    plt.grid(True)
    plt.legend()
    plt.show()

    plt.figure(figsize=(12, 4))
    plt.title(f"{title_prefix}: avg p_t")
    plt.plot(t, results["avg_p_cfr"], label="CFR avg p_t", linewidth=2)
    plt.plot(t, results["avg_p_eq"], label="Equilibrium avg p_t", linestyle="--", linewidth=2)
    plt.xlabel("t")
    plt.ylabel("p")
    plt.grid(True)
    plt.legend()
    plt.show()


def plot_mae_history(mae_hist: Dict[str, List[float]], title: str = "MC-CFR: MAE vs equilibrium over training"):
    if len(mae_hist["ep"]) == 0:
        print("[mae] No MAE logged (did you pass eq_for_logging and set log_every?)")
        return
    ep = np.array(mae_hist["ep"], dtype=int)
    mb = np.array(mae_hist["mae_beta"], dtype=float)
    ml = np.array(mae_hist["mae_lambda"], dtype=float)

    plt.figure(figsize=(12, 4))
    plt.title(title)
    plt.plot(ep, mb, label="MAE(beta)")
    plt.plot(ep, ml, label="MAE(lambda)")
    plt.xlabel("episode")
    plt.ylabel("MAE")
    plt.grid(True)
    plt.legend()
    plt.show()


# ============================================================
# Main
# ============================================================

if __name__ == "__main__":

    # -------------------------
    # 1) Tiny Kyle: Vanilla CFR
    # -------------------------
    print("=== Tiny Kyle: Vanilla CFR (exact discrete chance) ===")
    tiny_cfg = TinyKyleConfig(
        T=2,
        P0=0.5,
        V=1.0,
        N=1.0,
        X_ACTIONS=(-1.0, 0.0, 1.0),
        LAM_ACTIONS=(0.0, 0.5, 1.0),
        seed=0
    )
    tiny = TinyKyleVanillaCFR(tiny_cfg)
    tiny_avg_I, tiny_avg_M, tiny_hist, (I_track, M_track) = tiny.train(num_passes_over_chance=800)

    print(f"Tiny Kyle learned |I-infosets|={len(tiny_avg_I)}, |M-infosets|={len(tiny_avg_M)}")



    # -------------------------
    # 2) Large Kyle: MC-CFR (External Sampling)
    # -------------------------
    print("\n=== Large Kyle: MC-CFR (External Sampling) ===")
    P0 = 0.5
    SIGMA2_V = 2.0
    SIGMA2_U = 1.0
    SIGMA_V = float(np.sqrt(SIGMA2_V))
    SIGMA_U = float(np.sqrt(SIGMA2_U))
    T = 5

    eq = solve_kyle_model(T, SIGMA2_V, SIGMA2_U)
    print(f"[equilibrium] computed sequences for T={T} (scipy={_HAVE_SCIPY})")

    EPISODES = 5000
    LOG_EVERY = 1000

    INSIDER_ACTIONS = np.linspace(0.3, 2.0, 6)
    MM_ACTIONS = np.linspace(0.3, 2.0, 6)

    # Proactive experimentation hook:
    # Run once with False, once with True and compare MAE and plots.
    INCLUDE_P_IN_MM_INFOSET = False

    print(f"EPISODES={EPISODES}, T={T}")
    print(f"INSIDER_ACTIONS={INSIDER_ACTIONS}")
    print(f"MM_ACTIONS={MM_ACTIONS}")
    print(f"include_p_in_mm_infoset={INCLUDE_P_IN_MM_INFOSET}")

    mccfr = KyleExternalSamplingMCCFR(
        T=T,
        INSIDER_ACTIONS=INSIDER_ACTIONS,
        MM_ACTIONS=MM_ACTIONS,
        P0=P0,
        sigma_v=SIGMA_V,
        sigma_u=SIGMA_U,
        seed=0,
        snapshot_interval=None,
        snapshot_dir="cfr_snapshots",
        include_p_in_mm_infoset=INCLUDE_P_IN_MM_INFOSET
    )

    avg_I, avg_M, mae_hist = mccfr.train(
        num_episodes=EPISODES,
        log_every=LOG_EVERY,
        eq_for_logging=eq,
        eval_simulations=200,
        eval_seed=123
    )
    print("Training done.")

    # -------------------------
    # 3) Evaluation vs equilibrium (same as your later section)
    # -------------------------
    print("\n=== Evaluation vs multi-period equilibrium ===")
    results = evaluate_policies_vs_multi_period_equilibrium(
        avg_I=avg_I,
        avg_M=avg_M,
        T=T,
        INSIDER_ACTIONS=INSIDER_ACTIONS,
        MM_ACTIONS=MM_ACTIONS,
        P0=P0,
        sigma_v=SIGMA_V,
        sigma_u=SIGMA_U,
        eq=eq,
        num_simulations=500,
        seed=123,
        include_p_in_mm_infoset=INCLUDE_P_IN_MM_INFOSET
    )
    plot_eval(results, title_prefix="MC-CFR vs multi-period Kyle equilibrium")

    print("\nAll done.")
```
