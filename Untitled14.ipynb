```python
import numpy as np
import pettingzoo
import functools
from pettingzoo import AECEnv
from gymnasium import spaces
from pettingzoo.utils.agent_selector import AgentSelector
import math
from scipy.optimize import brentq
import matplotlib.pyplot as plt
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import matplotlib.pyplot as plt
from collections import defaultdict
```

```python
class KyleOnePeriodAEC(AECEnv):
    metadata = {
        "render_modes": ["human"],
        "name": "KyleOnePeriodAEC",
    }

    def __init__(self, sigma_v=2.0, sigma_u=1.0, p0=0.5, gamma=0.9, action_scaling=5, T=1):
        super().__init__()
        self.sigma_v_initial = sigma_v
        self.sigma_u_initial = sigma_u
        self.p0_initial = p0
        self.sigma_v = sigma_v
        self.sigma_u = sigma_u
        self.p0 = p0
        self.timestep = 0
        self.gamma = gamma
        self.T = T
        self.beta = np.zeros(self.T)
        self.lmbd = np.zeros(self.T)
        self.alpha = np.zeros(self.T)
        self.delta = np.zeros(self.T)
        self.Sigma = np.zeros(self.T)


        self.possible_agents = ["insider", "market_maker"]
        self.agents = self.possible_agents[:]
        self.render_mode = None
        self.rewards = {a: 0 for a in self.agents}
        self.agent_selector = AgentSelector(self.agents)

        self.action_spaces = {
            "insider": spaces.Box(low=-1, high=1, shape=(1,), dtype=np.float32),
            "market_maker": spaces.Box(low=-1, high=1, shape=(1,), dtype=np.float32)
        }

        self.observation_spaces = {
            "insider": spaces.Box(low=-10, high=10, shape=(3,), dtype=np.float32),
            "market_maker": spaces.Box(low=-10, high=10, shape=(3,), dtype=np.float32),
        }
        self.action_scaling = action_scaling

    def reset(self, seed=None, options=None):
        self.agents = self.possible_agents[:]
        self.rewards = {a: 0 for a in self.agents}
        self._cumulative_rewards = {a: 0 for a in self.agents}
        self.terminations = {a: False for a in self.agents}
        self.truncations = {a: False for a in self.agents}
        self.infos = {a: {} for a in self.agents}
        self.timestep = 0
        self.sigma_v = self.sigma_v_initial
        self.sigma_u = self.sigma_u_initial
        self.p0 = self.p0_initial
        self.optimal_x = 0
        self.optimal_y = 0
        self.beta = np.zeros(self.T)
        self.lmbd = np.zeros(self.T)
        self.alpha = np.zeros(self.T)
        self.delta = np.zeros(self.T)
        self.Sigma = np.zeros(self.T)

        # Sample true value
        self.v = torch.normal(self.p0, self.sigma_v, size=(1,)).item()
        self.u = torch.normal(0, self.sigma_u, size=(1,)).item()
        self.p = 0
        self.y = 0

        # Set first agent
        self.agent_selector.reinit(self.agents)
        self.agent_selection = self.agent_selector.next()

    def solve_kyle_model(self):
        """
        Calculates the multi-period Kyle model equilibrium using a two-pass algorithm.

        Args:
            N (int): The total number of trading periods.
            sigma2_v (float): The initial variance of the asset's value (Sigma_0).
            sigma2_u (float): The variance of the noise trader's orders.

        Returns:
            dict: A dictionary containing the numpy arrays for each equilibrium
                parameter sequence: 'Sigma', 'Lambda', 'Beta', and 'Alpha'.
        """
        # Helper for readability
        sigma_u = np.sqrt(self.sigma_u)

        # --- ⏪ PASS 1: BACKWARD ITERATION for h_n ---
        # h_n = Sigma_n / Sigma_{n-1}
        h = np.zeros(self.T)
        h[-1] = 0.5  # Terminal condition for the last period (n=N)

        # Iterate backwards from n = N-1 down to n=1
        for n in reversed(range(self.T - 1)):
            # Calculate theta_{n+1} from the known h_{n+1}
            h_next = h[n + 1]
            theta_next = np.sqrt(h_next / (1 - h_next))

            # Define the function whose root h_n we need to find
            # f(h_n) = theta_{n+1} - (2*h_n - 1) / (h_n * sqrt(1-h_n)) = 0
            def root_func(h_n):
                if h_n <= 0.5 or h_n >= 1.0: # Numerical stability
                    return np.inf
                return theta_next - (2 * h_n - 1) / (h_n * np.sqrt(1 - h_n))

            # Find the root h_n in the interval (0.5, 1.0)
            # brentq is a robust and efficient choice here.
            h[n] = brentq(root_func, 0.5 + 1e-9, 1.0 - 1e-9)

        # --- ⏩ PASS 2: FORWARD ITERATION for equilibrium parameters ---
        Sigma = np.zeros(self.T + 1)
        Lambda = np.zeros(self.T)
        Beta = np.zeros(self.T)
        Alpha = np.zeros(self.T)

        Sigma[0] = self.sigma_v_initial # Initial condition

        # Iterate forward from n=1 to N
        for n in range(self.T):
            # Update variance using the pre-calculated h_n
            Sigma[n + 1] = h[n] * Sigma[n]

            # Calculate the equilibrium parameters for period n
            Lambda[n] = np.sqrt((1 - h[n]) * Sigma[n + 1]) / sigma_u
            Beta[n] = (np.sqrt(1 - h[n]) * sigma_u) / np.sqrt(Sigma[n + 1])
            Alpha[n] = (sigma_u / (2 * np.sqrt(Sigma[n]))) * np.sqrt(h[n] / (1 - h[n]))

        self.beta = Beta
        self.lmbd = Lambda
        self.alpha = Alpha
        self.h = h
        self.Sigma = Sigma

        self.delta[self.T-1] = 0  # Terminal condition

        for n in reversed(range(1, self.T-1)):
            self.delta[n-1] = self.delta[n] + self.alpha[n] * ( self.lmbd[n] ** 2 ) * self.sigma_u


    def observe_insider(self):
        return torch.tensor([self.v, self.p0, self.beta[self.timestep]], dtype=torch.float32)

    def observe_market_maker(self):
        return torch.tensor([self.y, self.p0, self.lmbd[self.timestep]], dtype=torch.float32)

    def observe(self, agent):
        if agent == "insider":
            return self.observe_insider()
        elif agent == "market_maker":
            return self.observe_market_maker()

    @functools.lru_cache(maxsize=None)
    def observation_space(self, agent):
        return self.observation_spaces[agent]

    @functools.lru_cache(maxsize=None)
    def action_space(self, agent):
        return self.action_spaces[agent]

    def step(self, action):
        if self.terminations[self.agent_selection] or self.truncations[self.agent_selection]:
            return self._was_dead_step(action)

        agent = self.agent_selection

        # Store the action
        if agent == "insider":


            self.insider_action = action[0] * self.action_scaling
            self.u = torch.normal(0, self.sigma_u, size=(1,)).item()
            self.y = self.insider_action + self.u

        elif agent == "market_maker":
            self.market_maker_action = action[0] * self.action_scaling


            insider_profit = (self.v - self.market_maker_action) * self.insider_action
            market_maker_loss = - (self.market_maker_action - self.v) ** 2

            # Assign rewards
            self.rewards = {
                "insider": insider_profit,
                "market_maker": market_maker_loss
            }

            self.p0 = equilibrium_market_maker_action(self.y, self.p0, self.lmbd[self.timestep])

            # Update timestep
            self.timestep += 1
            if self.timestep >= self.T:
                self.terminations = {a: True for a in self.agents}

            # Update cumulative rewards
            for a in self.agents:
                self._cumulative_rewards[a] += self.rewards[a]

        # Get next agent
        self.agent_selection = self.agent_selector.next()

    def render(self):
        print(f"Step: {self.timestep}")
        print(f"True value v: {self.v:.2f}")
        print(f"Order flow y: {self.y:.2f}")
        if hasattr(self, 'rewards'):
            print(f"Agents' rewards: {self._cumulative_rewards}")

    def close(self):
        pass
```

```python
def solve_kyle_model(N: int, sigma2_v: float, sigma2_u: float):
    """
    Calculates the multi-period Kyle model equilibrium using a two-pass algorithm.

    Args:
        N (int): The total number of trading periods.
        sigma2_v (float): The initial variance of the asset's value (Sigma_0).
        sigma2_u (float): The variance of the noise trader's orders.

    Returns:
        dict: A dictionary containing the numpy arrays for each equilibrium
              parameter sequence: 'Sigma', 'Lambda', 'Beta', and 'Alpha'.
    """
    # Helper for readability
    sigma_u = np.sqrt(sigma2_u)

    # --- ⏪ PASS 1: BACKWARD ITERATION for h_n ---
    # h_n = Sigma_n / Sigma_{n-1}
    h = np.zeros(N)
    h[-1] = 0.5  # Terminal condition for the last period (n=N)

    # Iterate backwards from n = N-1 down to n=1
    for n in reversed(range(N - 1)):
        # Calculate theta_{n+1} from the known h_{n+1}
        h_next = h[n + 1]
        theta_next = np.sqrt(h_next / (1 - h_next))

        # Define the function whose root h_n we need to find
        # f(h_n) = theta_{n+1} - (2*h_n - 1) / (h_n * sqrt(1-h_n)) = 0
        def root_func(h_n):
            if h_n <= 0.5 or h_n >= 1.0: # Numerical stability
                return np.inf
            return theta_next - (2 * h_n - 1) / (h_n * np.sqrt(1 - h_n))

        # Find the root h_n in the interval (0.5, 1.0)
        # brentq is a robust and efficient choice here.
        h[n] = brentq(root_func, 0.5 + 1e-9, 1.0 - 1e-9)
# --- ⏩ PASS 2: FORWARD ITERATION for equilibrium parameters ---
    Sigma = np.zeros(N + 1)
    Lambda = np.zeros(N)
    Beta = np.zeros(N)
    Alpha = np.zeros(N)

    Sigma[0] = sigma2_v # Initial condition

    # Iterate forward from n=1 to N
    for n in range(N):
        # Update variance using the pre-calculated h_n
        Sigma[n + 1] = h[n] * Sigma[n]

        # Calculate the equilibrium parameters for period n
        Lambda[n] = np.sqrt((1 - h[n]) * Sigma[n + 1]) / sigma_u
        Beta[n] = (np.sqrt(1 - h[n]) * sigma_u) / np.sqrt(Sigma[n + 1])
        Alpha[n] = (sigma_u / (2 * np.sqrt(Sigma[n]))) * np.sqrt(h[n] / (1 - h[n]))

    return {
        "Sigma": Sigma,
        "Lambda": Lambda,
        "Beta": Beta,
        "Alpha": Alpha,
        "h": h
    }

if __name__ == '__main__':
    # --- Model Parameters ---
    N_periods = 25
    asset_value_variance = 2.0
    noise_trade_variance = 1.0

    # --- Solve the Model ---
    equilibrium = solve_kyle_model(N_periods, asset_value_variance, noise_trade_variance)

    # --- Display Results ---
    print("📈 Kyle Model Equilibrium Results")
    print(f"N = {N_periods}, Sigma_0 = {asset_value_variance}, sigma_u^2 = {noise_trade_variance}\n")

    # Create a pandas DataFrame for a clean table view
    periods = np.arange(1, N_periods + 1)
    df = pd.DataFrame({
        'Period': periods,
        'h_n': equilibrium['h'],
        'Sigma_n': equilibrium['Sigma'][1:],
        'Lambda_n': equilibrium['Lambda'],
        'Beta_n': equilibrium['Beta'],
        'Alpha_n': equilibrium['Alpha']
    })
    pd.set_option('display.precision', 4)
    print(df)

    # --- Plotting Key Dynamics ---
    fig, axs = plt.subplots(2, 2, figsize=(14, 10), constrained_layout=True)
    fig.suptitle(f'Kyle Model Equilibrium Dynamics (N={N_periods})', fontsize=16)

    # 1. Uncertainty (Sigma_n)
    axs[0, 0].plot(np.arange(N_periods + 1), equilibrium['Sigma'], 'o-', color='b')
    axs[0, 0].set_title('Market Maker Uncertainty ($\Sigma_n$)')
    axs[0, 0].set_xlabel('Period')
    axs[0, 0].set_ylabel('Variance')
    axs[0, 0].grid(True, linestyle='--', alpha=0.6)

    # 2. Price Impact (Lambda_n)
    axs[0, 1].plot(periods, equilibrium['Lambda'], 'o-', color='r')
    axs[0, 1].set_title('Price Impact ($\lambda_n$)')
    axs[0, 1].set_xlabel('Period')
    axs[0, 1].set_ylabel('Lambda')
    axs[0, 1].grid(True, linestyle='--', alpha=0.6)

    # 3. Trading Intensity (Beta_n)
    axs[1, 0].plot(periods, equilibrium['Beta'], 'o-', color='g')
    axs[1, 0].set_title('Insider Trading Intensity (β_n)')
    axs[1, 0].set_xlabel('Period')
    axs[1, 0].set_ylabel('Beta')
    axs[1, 0].grid(True, linestyle='--', alpha=0.6)

    # 4. Variance Reduction Rate (h_n)
    axs[1, 1].plot(periods, equilibrium['h'], 'o-', color='purple')
    axs[1, 1].set_title('Variance Reduction Rate ($h_n$)')
    axs[1, 1].set_xlabel('Period')
    axs[1, 1].set_ylabel('h_n')
    axs[1, 1].grid(True, linestyle='--', alpha=0.6)

    plt.show()
```

```python
def solve_kyle_equilibrium(T: int, sigma2_v: float, sigma2_u: float):
    """
    Multi-period Kyle equilibrium via two-pass algorithm.
    Returns dict with keys: 'Sigma', 'Lambda', 'Beta', 'Alpha', 'h'.
    """
    sigma_u = math.sqrt(sigma2_u)
    h = np.zeros(T)
    h[-1] = 0.5  # terminal boundary condition

    # Backward pass for h_n
    for n in range(T - 2, -1, -1):
        h_next = h[n + 1]
        theta_next = math.sqrt(h_next / (1.0 - h_next))

        def f(hn):
            if hn <= 0.5 or hn >= 1.0:
                return 1e6
            return theta_next - (2.0 * hn - 1.0) / (hn * math.sqrt(1.0 - hn))

        a, b = 0.5000001, 0.999999
        fa, fb = f(a), f(b)
        if fa * fb > 0:
            b = 0.95
            fa, fb = f(a), f(b)

        for _ in range(200):
            m = 0.5 * (a + b)
            fm = f(m)
            if abs(fm) < 1e-12 or (b - a) < 1e-12:
                h[n] = m
                break
            if fa * fm < 0:
                b, fb = m, fm
            else:
                a, fa = m, fm
        else:
            h[n] = 0.75  # fallback

    Sigma  = np.zeros(T + 1)
    Lambda = np.zeros(T)
    Beta   = np.zeros(T)
    Alpha  = np.zeros(T)
    Sigma[0] = sigma2_v
    eps = 1e-12

    for t in range(T):
        Sigma[t + 1] = h[t] * Sigma[t]
        Lambda[t] = math.sqrt(max(1.0 - h[t], 0.0) * Sigma[t + 1]) / sigma_u
        Beta[t]   = math.sqrt(max(1.0 - h[t], 0.0)) * sigma_u / math.sqrt(Sigma[t + 1] + eps)
        Alpha[t]  = (sigma_u / (2.0 * math.sqrt(Sigma[t] + eps))) * math.sqrt(h[t] / max(1.0 - h[t], eps))

    return {
        "Sigma": Sigma,
        "Lambda": Lambda,
        "Beta": Beta,
        "Alpha": Alpha,
        "h": h
    }
```

```python
# 1. Model & Game Configuration
# ============================================================
P0       = 0.5      # initial price
SIGMA_V  = 2.0      # std of fundamental value
SIGMA_U  = 1.0      # std of noise orders
NUM_PERIODS = 25    # number of periods per episode

# Discrete action grids (strategic parameters β, λ)
INSIDER_ACTIONS = np.linspace(0.0, 2.0, 9)   # β grid
MM_ACTIONS      = np.linspace(0.2, 0.6, 9)   # λ grid

# Theoretical equilibrium (for plotting)
eq = solve_kyle_equilibrium(NUM_PERIODS, SIGMA_V**2, SIGMA_U**2)
BETA_EQ   = eq["Beta"]     # β*_t
LAMBDA_EQ = eq["Lambda"]   # λ*_t

def kyle_one_period_payoffs(beta, lam, v, p_prev, u):
    """
    One-period Kyle dynamics + payoffs:
      x = β (v - p_prev)
      y = x + u
      p = p_prev + λ y

    Insider payoff: x (v - p)
    MM payoff:      -(p - v)^2
    """
    x = beta * (v - p_prev)
    y = x + u
    p = p_prev + lam * y

    insider_profit = x * (v - p)
    mm_profit      = -(p - v) ** 2
    return insider_profit, mm_profit, p

# Globals used so calculate_utility sees the same (v, p_{t-1}, u_t)
CURRENT_V      = None
CURRENT_P_PREV = None
CURRENT_U      = None

def set_current_state(v, p_prev, u):
    global CURRENT_V, CURRENT_P_PREV, CURRENT_U
    CURRENT_V      = v
    CURRENT_P_PREV = p_prev
    CURRENT_U      = u

def calculate_utility(agent_name, my_action, opponent_action):
    """
    Utility for a single turn based on Kyle model:
    - agent 'insider' action = β, opponent = λ
    - agent 'market_maker' action = λ, opponent = β

    Uses CURRENT_V, CURRENT_P_PREV, CURRENT_U set before this call.
    """
    if CURRENT_V is None:
        raise RuntimeError("State not set. Call set_current_state(...) before calculate_utility().")

    v      = CURRENT_V
    p_prev = CURRENT_P_PREV
    u      = CURRENT_U

    if agent_name == 'insider':
        beta = my_action
        lam  = opponent_action
        insider_profit, mm_profit, _ = kyle_one_period_payoffs(beta, lam, v, p_prev, u)
        return insider_profit
    else:  # market_maker
        beta = opponent_action
        lam  = my_action
        insider_profit, mm_profit, _ = kyle_one_period_payoffs(beta, lam, v, p_prev, u)
        return mm_profit
```

```python
# 2. Regret Matching Helpers
# ============================================================
def get_strategy(regret_sum):
    """
    regret_sum: vector of cumulative regrets for each action.
    Strategy is proportional to positive regrets, uniform if all <= 0.
    """
    positive_regrets = np.maximum(0.0, regret_sum)
    normalizing_sum = np.sum(positive_regrets)

    if normalizing_sum > 0:
        return positive_regrets / normalizing_sum
    else:
        return np.ones(len(regret_sum)) / len(regret_sum)

def sample_action(strategy, action_set):
    """Samples an action value from a strategy distribution."""
    action_index = np.random.choice(len(action_set), p=strategy)
    return action_set[action_index]

def expected_action_path(strategy_sum_agent, action_set, num_episodes_so_far):
    """
    Given strategy_sum_agent: (NUM_PERIODS, num_actions),
    compute E[action | period k] under the average strategy up to now.
    """
    avg_strat = strategy_sum_agent / max(num_episodes_so_far, 1)
    # (NUM_PERIODS, num_actions) @ (num_actions,) -> (NUM_PERIODS,)
    return avg_strat @ action_set
```

```python
# 3. Main Training Loop + snapshots for plotting
# ============================================================
def train_regret_matching(env:pettingzoo.AECEnv,
                          num_episodes: int,
                          snapshot_points=None):
    """
    Trains two agents in the Kyle model environment using per-state Regret Matching.

    - Actions: β from INSIDER_ACTIONS, λ from MM_ACTIONS
    - Utility: per-period payoff (profit/loss), summed over time.
    - No-regret: average regret per timestep -> 0.
    """
    if snapshot_points is None:
        snapshot_points = [10_000, 20_000, 30_000, 40_000, 50_000]

    agents = env.possible_agents  # ['insider', 'market_maker']
    action_spaces = {
        'insider': INSIDER_ACTIONS,
        'market_maker': MM_ACTIONS
    }
    num_actions = {name: len(spaces) for name, spaces in action_spaces.items()}

    # regrets[agent][period_k][action_idx]
    regret_sum = {
        agent: np.zeros((NUM_PERIODS, num_actions[agent])) for agent in agents
    }
    # strategy_sum for average strategy
    strategy_sum = {
        agent: np.zeros((NUM_PERIODS, num_actions[agent])) for agent in agents
    }
    # empirical joint action distribution for CCE
    empirical_dist = np.zeros((NUM_PERIODS,
                               num_actions['insider'],
                               num_actions['market_maker']))

    # For regret diagnostics
    regret_history = {agent: [] for agent in agents}

    # For plotting: snapshot of expected actions per period
    snapshots_insider = {}  # ep -> β_hat_t path (NUM_PERIODS,)
    snapshots_mm      = {}  # ep -> λ_hat_t path (NUM_PERIODS,)

    print(f"Starting training for {num_episodes} episodes...")

    for episode in range(num_episodes):
        ep_num = episode + 1
        if ep_num % 1000 == 0:
            print(f"  Episode {ep_num}/{num_episodes}")

        # Fresh value and initial price
        v = np.random.normal(loc=P0, scale=SIGMA_V)
        p_prev = P0

        # Noise sequence
        u_seq = np.random.normal(loc=0.0, scale=SIGMA_U, size=NUM_PERIODS)

        # --- Run a single episode ---
        for k in range(NUM_PERIODS):
            u_k = u_seq[k]

            # Set state for this period (used inside calculate_utility)
            set_current_state(v, p_prev, u_k)

            # Current strategies for period k
            strategies = {
                agent: get_strategy(regret_sum[agent][k]) for agent in agents
            }

            # Actions drawn from mixed strategies
            insider_action = sample_action(strategies['insider'], action_spaces['insider'])
            mm_action      = sample_action(strategies['market_maker'], action_spaces['market_maker'])

            # Update empirical distribution
            insider_action_idx = np.where(action_spaces['insider'] == insider_action)[0][0]
            mm_action_idx      = np.where(action_spaces['market_maker'] == mm_action)[0][0]
            empirical_dist[k, insider_action_idx, mm_action_idx] += 1

            # Kyle dynamics for this period (to evolve price)
            beta = insider_action
            lam  = mm_action
            insider_pi, mm_pi, p_next = kyle_one_period_payoffs(beta, lam, v, p_prev, u_k)

            # Regret updates
            for agent in agents:
                my_action = insider_action if agent == 'insider' else mm_action
                opp_action = mm_action if agent == 'insider' else insider_action

                actual_utility = calculate_utility(agent, my_action, opp_action)

                for i, possible_action in enumerate(action_spaces[agent]):
                    counterfactual_utility = calculate_utility(agent, possible_action, opp_action)
                    regret = counterfactual_utility - actual_utility
                    regret_sum[agent][k, i] += regret

            # Strategy sum accumulation (for average strategy)
            for agent in agents:
                strategy_sum[agent][k] += strategies[agent]

            # evolve price
            p_prev = p_next

        # --- regret_at_T / T diagnostics ---
        T_so_far = ep_num
        for agent in agents:
            period_regrets = np.max(regret_sum[agent], axis=1)  # (NUM_PERIODS,)
            avg_regret_per_timestep = np.mean(period_regrets) / T_so_far
            regret_history[agent].append(avg_regret_per_timestep)

        # --- snapshots for plotting learned actions vs equilibrium ---
        if ep_num in snapshot_points:
            print(f"Taking snapshot at episode {ep_num}...")
            snapshots_insider[ep_num] = expected_action_path(
                strategy_sum["insider"], INSIDER_ACTIONS, ep_num
            )
            snapshots_mm[ep_num] = expected_action_path(
                strategy_sum["market_maker"], MM_ACTIONS, ep_num
            )

    print("Training finished.")

    # Final CCE & average strategies
    total_plays_per_period = np.sum(empirical_dist, axis=(1, 2), keepdims=True)
    cce_distribution = empirical_dist / np.maximum(total_plays_per_period, 1.0)

    avg_strategies = {
        agent: strategy_sum[agent] / np.sum(strategy_sum[agent], axis=1, keepdims=True)
        for agent in agents
    }

    return cce_distribution, avg_strategies, regret_history, snapshots_insider, snapshots_mm
```

```python
def plot_action_convergence(snapshots_insider, snapshots_mm):
    """
    Produce 10 plots total:
      - 5 for insider: learned x_t vs equilibrium x_t*
      - 5 for MM: learned p_t vs equilibrium p_t*
    """

    # Use same random v and u_t for both EQ and learned paths
    v = np.random.normal(loc=P0, scale=SIGMA_V)
    u_seq = np.random.normal(loc=0.0, scale=SIGMA_U, size=NUM_PERIODS)

    for ep in sorted(snapshots_insider.keys()):
        beta_hat  = snapshots_insider[ep]     # (NUM_PERIODS,)
        lambda_hat = snapshots_mm[ep]         # (NUM_PERIODS,)

        # ---------------------------------------------
        # Compute learned x_t and p_t paths
        # ---------------------------------------------
        p_rl = np.zeros(NUM_PERIODS + 1)
        p_rl[0] = P0
        x_rl = np.zeros(NUM_PERIODS)
        y_rl = np.zeros(NUM_PERIODS)

        for t in range(NUM_PERIODS):
            x_rl[t] = beta_hat[t] * (v - p_rl[t-1])              # insider action
            y_rl[t] = x_rl[t] + u_seq[t]                       # OF
            p_rl[t] = p_rl[t-1] + lambda_hat[t] * y_rl[t]      # MM price

        # ---------------------------------------------
        # Compute equilibrium x_t* and p_t* paths
        # ---------------------------------------------
        p_eq = np.zeros(NUM_PERIODS + 1)
        p_eq[0] = P0
        x_eq = np.zeros(NUM_PERIODS)
        y_eq = np.zeros(NUM_PERIODS)

        for t in range(NUM_PERIODS):
            x_eq[t] = BETA_EQ[t] * (v - p_eq[t-1])
            y_eq[t] = x_eq[t] + u_seq[t]
            p_eq[t] = p_eq[t-1] + LAMBDA_EQ[t] * y_eq[t]

        periods = np.arange(1, NUM_PERIODS + 1)
        periods0 = np.arange(0, NUM_PERIODS + 1)

        # ---------------------------------------------
        # Plot insider x_t
        # ---------------------------------------------
        plt.figure(figsize=(10,4))
        plt.plot(periods, x_eq, "k--", label="Equilibrium $x_t^*$")
        plt.plot(periods, x_rl, "o-", label=fr"Learned $x_t$ at {ep} episodes")
        plt.xlabel("t")
        plt.ylabel("$x_t$ (insider trading order)")
        plt.title(f"Insider actions vs equilibrium (episode {ep})")
        plt.grid(alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

        # ---------------------------------------------
        # Plot MM p_t
        # ---------------------------------------------
        plt.figure(figsize=(10,4))
        plt.plot(periods0, p_eq, "k--", label="Equilibrium $p_t^*$")
        plt.plot(periods0, p_rl, "o-", label=fr"Learned $p_t$ at {ep} episodes")
        plt.xlabel("t")
        plt.ylabel("$p_t$ (market maker price)")
        plt.title(f"Market maker prices vs equilibrium (episode {ep})")
        plt.grid(alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()
```
